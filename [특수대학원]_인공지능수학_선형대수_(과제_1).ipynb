## 행렬 정의

다음의 3x3 행렬 $A$를 정의하는 numpy 코드를 작성하시오.

$$ A =
\left[
\begin{array}{rrr}
 1 &  2 &  1 \\
 1 &  2 &  3 \\
 2 &  3 & -1
\end{array}
\right]
$$

```
import numpy as np
import numpy.linalg
```

```
# 행렬 코딩

A = np.array([[1, 2, 1], [1, 2, 3], [2, 3, -1]])

print(A)
print(np.shape(A))
```

## 벡터 정의
다음의 3-벡터 $\mathbf{b}$를 정의하는 numpy 코드를 작성하시오.
$$ \mathbf{b} =
\left[
\begin{array}{rrr}
 1 \\
 3 \\
-3
\end{array}
\right]
$$

```
# 벡터 코딩

## TODO
b = np.array([[1],[3],[-3]])

print(b)
print(np.shape(b))
```

## 역행렬 구하기

3x3 행렬 $A$의 역행렬을 구하는 코드를 numpy의 linalg 서브모듈의 inv() 함수를 이용해 작성하시오.

$$ A^{-1} =
\left[
\begin{array}{rrr}
 1 &  2 &  1 \\
 1 &  2 &  3 \\
 2 &  3 & -1
\end{array}
\right]^{-1}
$$

참고자료: [링크](https://numpy.org/doc/stable/reference/generated/numpy.linalg.inv.html)

```
# 역행렬 구하기

## TODO
A_inv =  np.linalg.inv(A)
# check inv correct
print(A@A_inv)

print(A_inv)
print(np.shape(A_inv))
```
